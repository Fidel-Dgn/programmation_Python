#### Exercice 1.2 : parcours d'un tableau à 2 dimensions

Effectuez les transformations suivantes. 

Dans des cellules de code séparées : 

1) créer image_negatif en inversant les valeurs 0 <--> 255 et afficher l'image pour vérifier ainsi que le nombre de pixel noir et blanc après transformation

2) créer une image en changeant la couleur de fond (blanc) par un niveau de gris. Contrôler la validité de la valeur lue avant d'effectuer le traitement. Sortir au bout de 3 essais sinon. Conserver la valeur BLANC sur tout le pourtour de la figure.

3) Pour chaque transformation, vous créerez une copie de l'image d'origine lue et vous vérifierez que l'identifiant de la copie est bien différent de celui de l'original et vous afficherez les dimension de l'image traitée. 


    



```python
# Exercice 1.2 - Image en négatif
image_negatif=mon_image_nb.copy()
for i in range(hauteur):
    for j in range(largeur):
        if image_negatif[i][j] == 0:
            image_negatif[i][j] = 255
        else:
            image_negatif[i][j] = 0
nb_pixels_noir = np.sum(image_negatif == 0)
nb_pixels_blanc = np.sum(image_negatif == 255)
print('Nombre de pixels noirs = ', nb_pixels_noir, '\n')
print('Nombre de pixels blancs = ', nb_pixels_blanc, '\n')
print("id_mon_image_nb:", id(mon_image_nb), '\n')
print("id_image_negatif:", id(image_negatif), '\n')
print('Taille de l\'image = ', image_negatif.shape[0], ' x ', image_negatif.shape[1], '\n')

plt.imshow(image_negatif, cmap='gray')  
     
```

#### Exercice 2 : image et boite englobante

A partir de l'image noir et blanc initiale, tracer un cadre en gris autour de l'objet en noir. 

Méthode : Pour avoir la boite englobante d'un objet détecté il faut avoir le coin supérieur gauche et le coin inférieur droit. 
Prendre une marge de DELTA pixels (à gauche et à droite, en haut et en bas, pour que le cadre ne colle pas à l'objet.

```python
### EXERCICE 2 : Manipulation de tableau à 2 dimensions  


# Sauvergarde de la nouvelle image incluant la boite englobante
PATH_SAUVEGARDE = 'C:/Users/Etudiant/Documents/fidel/Projet/programmation_Python/ETUD_TP2_SUJETS/OBJECT_DETECTE' 


for i in range(51, 64):
    mon_image = mpl.image.imread(PATH + f'/{i}.bmp')
    
    # Compléter le code
    img_boite_englobante = mon_image.copy()
    delta=10  # Définir la valeur de delta selon vos besoins
    Gray_value = 128  # Valeur de gris pour le fond
    ligne_noire, colonne_noire = np.where(img_boite_englobante == 0)

    print("Lignes des pixels noirs : ", ligne_noire)
    print("Colonnes des pixels noirs : ", colonne_noire)
    if (len(ligne_noire)==0) or (len(colonne_noire)==0):
        print("Aucun objet détecté donc pas de boîte englobante trouvée.")
    else:
        ligne_min = np.min(ligne_noire)
        ligne_max = np.max(ligne_noire)
        colonne_min = np.min(colonne_noire)
        colonne_max = np.max(colonne_noire)

    # Ajuster les coordonnées de la boîte englobante avec delta
        ligne_min_ajust = max(ligne_min - delta, 0)
        ligne_max_ajust = min(ligne_max + delta, img_boite_englobante.shape[0] - 1)
        colonne_min_ajust = max(colonne_min - delta, 0)
        colonne_max_ajust = min(colonne_max + delta, img_boite_englobante.shape[1] - 1)

    #trace des lignes de la boîte englobante sur l'image
        img_boite_englobante[ligne_min_ajust:ligne_max_ajust+1, colonne_min_ajust] = Gray_value  # Ligne gauche
        img_boite_englobante[ligne_min_ajust:ligne_max_ajust+1, colonne_max_ajust] = Gray_value  # Ligne droite
        img_boite_englobante[ligne_min_ajust, colonne_min_ajust:colonne_max_ajust+1] = Gray_value  # Ligne haut
        img_boite_englobante[ligne_max_ajust, colonne_min_ajust:colonne_max_ajust+1] = Gray_value  # Ligne bas
        

    

    plt.figure()
    plt.imshow(img_boite_englobante, cmap='gray')  
    plt.savefig(PATH_SAUVEGARDE + f'/image{i}.jpg')
        
```
