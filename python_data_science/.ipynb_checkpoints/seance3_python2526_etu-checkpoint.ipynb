```python
feature2plot = 'bill_length_mm' # ont change les type de pinguin a chaque fois 
plt.hist(data[data['species'] == 'Adelie'][feature2plot],
         bins=30,
         alpha=0.6,
         density=True,
         label='Hommes')

plt.hist(data[data['species'] == 'Chinstrap'][feature2plot],
         bins=30,
         alpha=0.6,
         density=True,
         label='Femmes')

plt.title('Distribution des âges selon le sexe')
plt.xlabel('Âge')
plt.ylabel('Proportion')
plt.legend()

plt.show()
```
