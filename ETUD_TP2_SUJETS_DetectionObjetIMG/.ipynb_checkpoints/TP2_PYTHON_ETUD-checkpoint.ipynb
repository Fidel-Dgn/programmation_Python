#### Exercice 1.1
Exécuter le code suivant, en remplaçant la valeur de la variable PATH par le chemin vers le dossier récupéré sur moodle et rangé dans le dossier dédié au TP2 (dossier TP2_PYTHON_NOM_ETUD à créer sous TP_PYTHON). 

Etudier le résultat et commenter. 

```python
# Exercice 1.1

# bibliothèque à importer

from PIL import Image              # pour la manipulation d'images
import numpy as np                 # pour la gestion des tableaux 

import matplotlib as mp            # pour tracer des figures
from matplotlib import pyplot      

# mémorisation du chemin vers le dossier contenant les données à traiter
PATH = '/Users/ferrane/Documents/ENSEIGNEMENT/SRI-1A/FIL_ROUGE_2023/DATA_FIL_ROUGE_DEV'

# chagement d'une image donnée noir et blanc - résultant d'une binarisation
image_lue = Image.open(PATH + '/IMG_et_AUDIO/TEST_NB/51.bmp')

# affichage de l'image  
print(type(image_lue), '\n')
print(image_lue)
image_lue.show()

# récupération de l'image lue dans un tableau numpy et de ses dimensions hauteur, largeur

mon_image_nb = np.array(image_lue)
hauteur, largeur = mon_image_nb.shape

print('Taille de l\'image = ', hauteur, ' x ', largeur, '\n')
print('type tableau = ', type(mon_image_nb), 'type élément = ', type(mon_image_nb[0][0]), '\n')
print(mon_image_nb)
pyplot.imshow(mon_image_nb, cmap='gray')
```
