Lors de la récupération d'un dataset, le premier travail du *data scientist* consiste à visualiser, trier et préparer les données, puis à vérifier qu'il n'y a pas d'erreur. Dans cette partie, nous allons explorer les données pour préparer le dataset à une exploitation statistique.

Si vous reprenez la ligne numéro 1 ci-dessus, vous verrez par exemple que dans la colonne `Cabin` apparaît la mention `NaN`. Ce n'est pas une chaîne de caractères : c'est un type Python signifiant *Not a Number*, c'est-à-dire une valeur absente. Nous avons également vu que certains passagers n'ont pas renseigné leur âge ; il y a donc aussi des `NaN` dans la colonne `Age`.

Il se peut également qu'il existe des valeurs aberrantes dans le dataset. Dans le tableau des statistiques, vous avez peut-être remarqué que la variable `Age` a pour minimum -3 ans et pour maximum 138 ans.

Enfin, certaines variables peuvent être dans un format peu pratique pour les calculs, ou présenter une précision inutile. La variable `Fare` sera arrondie au centime et la variable `Survived` sera convertie en `bool`. Les colonnes inutiles seront également supprimées.

<h2>Suppression des données inutiles</h2>

Pour réduire la taille du dataset, il est possible de supprimer directement les colonnes qui ne seront pas utiles par la suite. C'est le cas des colonnes `Ticket`, `Cabin` et `Name`. Pour supprimer une colonne, on utilise la méthode [`.drop(columns='NomColonne')`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop.html).

```python
data.drop(columns=['Ticket', 'Cabin', 'Name'], inplace=True)
```

<h2>Changement de type</h2>

Attention à ne pas confondre les types propres aux langages de programmation (`int`, `float`, `char`) et les types de variables liés à l'analyse de données. En *data science* on distingue deux grandes familles de variables :
 - les **variables catégorielles** (ou qualitatives) : elles ne représentent pas des valeurs quantifiables mais des catégories distinctes et mutuellement exclusives. C'est le cas de `Pclass`, car on ne peut appartenir qu'à une seule classe et calculer une moyenne sur les classes n'a pas de sens. Ces variables peuvent être encodées par des entiers, des chaînes de caractères, des booléens, etc. ;
 - les **variables numériques** (ou quantitatives) : elles représentent des grandeurs mesurables (par exemple `Age`). Elles sont toujours de type numérique sur la machine (`int`, `float`).

Pour changer le type des valeurs d'une colonne, il faut sélectionner la colonne puis utiliser la méthode [`.astype('type')`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.astype.html).

La variable `Survived` est de type `int64` ; nous allons la convertir en `bool`, qui occupe moins de place en mémoire.

```python
data['Survived']=data['Survived'].astype('bool')
```

La variable `Pclass` est représentée par des nombres, mais effectuer des calculs numériques avec elle n'a pas de sens : c'est une variable qualitative. Il est possible de l'indiquer à Pandas en spécifiant le type `category`.

```python
data['Pclass']=data['Pclass'].astype('category')
```

```python
data.describe()
```

<h2>Gestion des données aberrantes</h2>

À la lecture des statistiques ci-dessus, on remarque que certains passagers ont des âges impossibles (-3 ans) ou très peu probables (138 ans). Pour l'âge négatif, nous supprimerons ce passager (s'il a menti sur son âge, ses autres données sont probablement peu fiables). Pour retrouver un passager dans la liste, on peut utiliser des conditions dans l'indexage ; l'attribut `.index` permet d'obtenir son identifiant.

```python
data.drop(index=data[data['Age']==-3].index,inplace=True)
data.describe()
```

Pour la personne de 138 ans, bien que ce soit peu probable, ce n'est pas strictement impossible. Regardons si c'est un cas isolé ou si plusieurs passagers sont très âgés, grâce à la méthode [`.sort_values()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.sort_values.html).

```python
data['Age'].sort_values(ascending=False)[0:10]
```

Au vu des données, la personne de 138 ans est bien un cas isolé. Nous choisissons de la supprimer (des analyses statistiques complémentaires sur la distribution des âges pourraient également justifier ce choix).

```python
data.drop(index=data[data['Age']==138].index,inplace=True)
data.describe()
```

<h2>Arrondis</h2>

La colonne `Fare` représente le prix du billet et sera arrondie au centime le plus proche. Pour cela, il faut sélectionner la colonne à modifier et utiliser la méthode [`.round(nbDecimale)`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.round.html). De même pour la variable `LWeight`, arrondie à 100 g près.

```python
data['Fare'] = data['Fare'].round(2)
data['LWeight'] = data['LWeight'].round(1)
```

```python
data.head()
```

<h2>Gestion des données manquantes</h2>

Nous avons maintenant un dataset allégé et partiellement nettoyé. Il reste à gérer les données manquantes, notamment dans la variable `Age` où certaines valeurs sont notées `NaN`. Pour les identifier, on utilise la méthode [`.isna()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.isna.html) ; pour les compter, on enchaîne avec [`.sum()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sum.html).

```python
data['Age'].isna().sum()
```

Il y a donc 176 valeurs manquantes concernant l'âge des passagers.

Plusieurs stratégies existent pour gérer ces données manquantes. On pourrait supprimer les passagers dont l'âge est inconnu, mais ils sont nombreux et cette approche réduirait trop le dataset. Une approche plus prudente consiste à remplacer les valeurs manquantes par un indicateur qui perturbe le moins possible les statistiques : par exemple la médiane [`.median()`](https://pandas.pydata.org/docs/reference/api/pandas.Series.median.html) ou la moyenne [`.mean()`](https://pandas.pydata.org/docs/reference/api/pandas.Series.mean.html) des âges connus.

Avant cela, vérifions si l'âge médian diffère selon le sexe des passagers en regroupant les données avec [`.groupby('Sex')`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html) :

```python
median_ages=data.groupby('Sex')['Age'].median()
median_ages
```

L'âge médian étant différent selon le sexe, nous remplaçons les valeurs manquantes par la médiane correspondant au sexe du passager, ce qui donne une estimation plus précise. Pour ce faire, nous utiliserons la méthode [`pandas.DataFrame.apply`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html) et une fonction [`lambda`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html), c’est-à-dire une fonction anonyme permettant de définir une opération simple en une seule ligne, particulièrement utile dans ce type de transformation de données.

```python
data['Age']=data.apply(lambda x: median_ages[x['Sex']] if pd.isna(x['Age']) else x['Age'],axis=1)
```

```python
data.head()
```

Concernant la variable `Embarked`, il manque deux valeurs. Nous pouvons les remplacer par la valeur la plus fréquente. Comme il ne s'agit pas d'une variable numérique, il n'est pas possible de calculer une moyenne ou une médiane ; en revanche, la méthode [`.mode()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.mode.html) identifie la valeur la plus présente. Pour remplacer tous les `NaN` par une valeur donnée, on utilise [`.fillna(valeur)`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.fillna.html).

```python
mostCommonEmb=data['Embarked'].mode()[0]
data['Embarked']=data['Embarked'].fillna(mostCommonEmb)
```

<h3>Histogramme</h3>

Le premier graphique intéressant est la distribution des âges parmi les passagers. Un histogramme [`plt.hist()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html) est la représentation classique pour visualiser la distribution d'une variable numérique.

```python
plt.hist(data['Age'], bins=30)
plt.title('Distribution des âges')
plt.xlabel('Âge')
plt.ylabel('Nombre de passagers')
plt.show()
```

On remarque sur le graphique précédent une forte proportion de passagers ayant entre 27 et 30 ans. Ceci est normal, car nous avons remplacé les âges manquants par la médiane des âges, qui était respectivement de 27 et 29 ans selon le sexe.

Il est également possible de représenter la distribution des âges selon le sexe sous forme d’histogrammes normalisés (en pourcentage) en utilisant l’option density=True.

```python
plt.hist(data[data['Sex'] == 'male']['Age'],
         bins=30,
         alpha=0.6,
         density=True,
         label='Hommes')

plt.hist(data[data['Sex'] == 'female']['Age'],
         bins=30,
         alpha=0.6,
         density=True,
         label='Femmes')

plt.title('Distribution des âges selon le sexe')
plt.xlabel('Âge')
plt.ylabel('Proportion')
plt.legend()

plt.show()
```

<h3>Diagramme à barres</h3>

Afin d’analyser l’influence du sexe sur les chances de survie, il est pertinent d’étudier la répartition des survivants et des non-survivants selon cette variable.

Le diagramme à barres groupées est particulièrement adapté à la visualisation de variables catégorielles, car il permet de comparer facilement plusieurs catégories entre elles.

Nous allons calculer le nombre de passagers pour chaque combinaison des variables Survived et Sex à l’aide de la fonction [`pandas.crosstab()`](https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html). Le tableau (DataFrame) obtenu sera ensuite représenté graphiquement grâce à la méthode .plot() afin d’obtenir un diagramme à barres illustrant la répartition des survivants selon le sexe.

```python
counts = pd.crosstab(data['Survived'], data['Sex'])
counts
```

```python
counts.plot(kind='bar')
plt.title('Répartition des survivants par sexe')
plt.xlabel('Survivant')
plt.ylabel('Nombre')
plt.legend(title='Sexe')
plt.show()
```

Il est aussi possible d'afficher ce graphique en pourcentage pour faciliter les comparaisons entre groupes de taille différente.

```python
counts_norm = pd.crosstab(data['Survived'], data['Sex'], normalize='all') * 100

counts_norm.plot(kind='bar')
plt.title('Répartition des survivants par sexe')
plt.xlabel('Survivant')
plt.ylabel('% du nombre de passager total')
plt.legend(title='Sexe')
plt.show()
```

Pour cela, vous utiliserez une fonction [`lambda`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html), c’est-à-dire une fonction anonyme permettant de définir une opération simple en une seule ligne, particulièrement utile dans ce type de transformation de données. À partir de cette variable, construisez une table permettant d’analyser le taux de survie selon ces différents groupes, puis représentez ces résultats sous forme de diagramme en barres afin de visualiser les différences de survie entre les catégories.

<h3>Nuage de points</h3>

Lorsque nous avons deux variables numériques, il est possible de représenter l'une en fonction de l'autre sous forme de nuage de points avec la fonction [`plt.scatter()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html). Commençons par observer le prix du billet en fonction de l'âge, en colorant les points selon la survie du passager.

Cette fois-ci, une tendance se dessine clairement : plus le poids des bagages augmente, plus le prix du billet augmente. Cela semble une hypothèse raisonnable, car le prix du billet était souvent proportionnel au poids des bagages autorisés.

<h3>Statistiques intermédiaires</h3>

Pour mettre en évidence des tendances, il est utile de calculer des statistiques intermédiaires avant l'affichage, afin de regrouper les données ou de calculer des indicateurs synthétiques.

Par exemple, on peut calculer le taux de survie (moyenne des survivants) puis l'afficher en fonction du nombre de membres de la famille à bord. Le taux de survie devient ainsi un indicateur calculé à partir d'une variable catégorielle.

```python
survival_rate=data.groupby('SibSp')['Survived'].mean()

plt.figure()
plt.scatter(survival_rate.index,survival_rate.values)

plt.title('Taux de survie en fonction de SibSp')
plt.xlabel('Nombre de frere/soeurs/conjoint a bord de (SibSp)')
plt.ylabel('Poids des baggages')
plt.ylabel('Taux de surface')
plt.ylim(0,1)
plt.show()
```

Dans l'autre sens, il est possible de créer une variable catégorielle à partir d'une variable quantitative pour regrouper les données dans des classes plus faciles à étudier.

Commençons par regrouper les prix des billets en dix classes (les déciles) pour cela on utilisera la fonction [`pandas.qcut()`](https://pandas.pydata.org/docs/reference/api/pandas.qcut.html). Puis calculons le taux moyen de survie et la variance du taux de survie par décile avec la fonction [`pandas.DataFrame.agg`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.agg.html). Finalement affichons le résultat sous forme de courbe avec barres d'erreur avec la fonction [`matplotlib.pyplot.errorbar`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.errorbar.html). 

```python
data['Fare_decile'] = pd.qcut(data['Fare'], q=10)

grp = data.groupby('Fare_decile')['Survived' ].agg(['mean', 'std'])
plt.figure()
plt.errorbar(grp.index.astype(str), grp['mean'], yerr=grp['std'])
plt.xlabel('Décile de prix')
plt.ylabel('Taux de survie moyen')
plt.xticks(rotation=45)

plt.show()
```

<center><h2>Corrélation</h2></center>

La corrélation permet de quantifier le lien linéaire entre deux variables quantitatives.

<p align="center">
  <img src="Correlation_examples2.svg" width="800"/><br>
  <em>
    Nuages de points et leurs coefficients de corrélation linéaire, source : <a href="https://fr.wikipedia.org/wiki/Corr%C3%A9lation_(statistiques)">Wikipédia</a>
  </em>
</p>

Pour calculer l'indice de corrélation, il faut sélectionner les deux colonnes à étudier puis utiliser la méthode [`.corr()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.corr.html).

D'après le nuage de points entre l'âge et le prix du billet, il ne semble pas y avoir de lien direct entre les deux variables. Calculons l'indice de corrélation pour le confirmer.

```python
print(data[['Age','Fare']].corr())
```

L'indice de corrélation entre les deux variables est proche de 0, ce qui confirme l'absence de lien linéaire entre l'âge et le prix du billet.

Effectuons le même calcul entre le prix du billet et le poids des bagages.

```python
print(data[['Fare','LWeight']].corr())
```

Cette fois-ci, l'indice de corrélation est proche de 1, ce qui confirme l'existence d'un lien fort entre les deux variables — cohérent avec ce que nous avions observé sur le nuage de points.

<div class="alert alert-block alert-danger"><center>La corrélation entre deux variables ne signifie pas qu’il existe un lien de causalité entre elles ! </center></div>

De nombreux exemples amusants mais trompeurs sont disponibles sur le site [Spurious Correlations](https://www.tylervigen.com/spurious-correlations), comme l’illustre l’image ci-dessous montrant une corrélation entre la consommation de margarine et le taux de divorce :

<p align="center">
  <img src="Margarine_vs_divorce.png" width="800"/>
</p>


Ces corrélations sont dites *fallacieuses* car elles sont souvent dues à un **facteur de confusion**, c’est-à-dire une troisième variable non observée qui influence simultanément les deux variables étudiées. Comme expliqué sur la page Wikipédia consacrée au sujet ([Corrélation n'implique pas causalité](https://fr.wikipedia.org/wiki/Corr%C3%A9lation_n%27implique_pas_causalit%C3%A9)) :

<p align="center"><em>
Une étude scientifique annonce : « Les jeunes enfants qui dorment avec une veilleuse ont plus de chance de devenir myopes plus tard ». Cette étude de l'université de Pennsylvanie a été publiée le 13 mai 1999 dans la revue Nature et a fait l'objet d'une couverture médiatique. Cependant, une autre étude de l'université de l'Ohio réalisée plus tard n'a trouvé aucun lien entre le développement de la myopie et le fait de dormir la lumière allumée ; en revanche, elle montre un lien important entre la myopie des parents et celle de leurs enfants et remarque que les parents myopes ont tendance à laisser une lumière allumée la nuit pour leurs enfants.

À partir du dataset nettoyé et en vous appuyant sur des graphiques, essayez de répondre à la question du taux de survie de Jack et de Rose en fonction des informations connues les concernant.

```python
# Groupe "Jack" : Hommes de la 3e classe
jack_group = data[(data['Sex'] == 'male') & (data['Pclass'] == 3)]

# Groupe "Rose" : Femmes de la 1ère classe
rose_group = data[(data['Sex'] == 'female') & (data['Pclass'] == 1)]

# Fonction de moyenne glissante
def moving_average(series, window=5):
    values = series.values
    result = []
    half = window // 2

    for i in range(len(values)):
        start = max(0, i - half)
        end = min(len(values), i + half + 1)
        result.append(values[start:end].mean())

    return pd.Series(result, index=series.index)

# Calcul du taux de survie par âge
jack_mean = (
    jack_group.sort_values('Age')
              .groupby('Age')['Survived']
              .mean()
)

rose_mean = (
    rose_group.sort_values('Age')
              .groupby('Age')['Survived']
              .mean()
)

# Lissage
survie_jack_age = moving_average(jack_mean, window=5)
survie_rose_age = moving_average(rose_mean, window=5)

# Tracé
plt.figure()
plt.plot(survie_jack_age.index, survie_jack_age.values, label='Hommes, Classe 3 (Jack)')
plt.plot(survie_rose_age.index, survie_rose_age.values, label='Femmes, Classe 1 (Rose)')
plt.title("Évolution du taux de survie selon l'âge (Moyenne glissante 5)")
plt.xlabel('Âge')
plt.ylabel('Taux de survie')
plt.legend()
plt.show()
```
