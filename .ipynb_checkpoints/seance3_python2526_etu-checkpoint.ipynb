## 1. Analyse de la base de données

```python
# Ouverture du fichier
data = pd.read_csv('penguins.csv')
```

```python
data
```

```python
data.describe()
#pas de valeur aberantes que des neantes Nan
```

```python
data["species"].unique()
```

```python
print(sum(data["species"]=='Adelie'))
print(sum(data["species"]=='Chinstrap'))
print(sum(data["species"]=='Gentoo'))
```

```python
data_bin=data[(data["species"]=='Adelie')| (data["species"]=='Chinstrap')]

data_bin['species'].unique()
```

```python
data_bin
```

```python
feature2plot=['bill_length_mm','bill_depth_mm','flipper_length_mm','body_mass_g']
for feature in feature2plot:
    for species in ['Adelie','Chinstrap']:
        plt.hist(data_bin[data_bin['species']==species][feature],bins=30,alpha=0.6,label=species)
    plt.xlabel(feature)
    plt.ylabel('count')
    plt.legend()
    plt.show()
```

```python
#"filtrage
data_bin_2features=data_bin[['species','bill_length_mm']]
data_bin_2features
```

```python
#supression de NaN
data_bin_2feature_filtre=data_bin_2features.dropna()
data_bin_2feature_filtre
```

```python
#3. regressiion logistique
```

```python
x=data_bin_2feature_filtre[['bill_length_mm']]
print("X shape:",x.shape)
print("X type:",type(x))
```

```python
# On extrait La colonne "bill _Length mm
# ATTENTION # ['bill_Length_mm']-> retourne une series (1D)
#[["bill_Length_mm']] -> retourne un DataFrame (20)
# En Machine Learning (scikit-Learn), 
# X doit avoir la forme : (n_samples, n_features) donc on utilise DEUX crochets,

X= data_bin_2feature_filtre[['bill_length_mm']]
#Affiche les dimensions de x # Format ; (nombre de_Lignes, nombre_de_colonnes) 
print("* shape:", X.shape)
# verifier le type de la variable 
# icic on obtient :(class 'pandas.core.frame.dataframe 
print("X type:",type(X))
```

```python
data_bin_2feature_filtre['species']=data_bin_2feature_filtre['species'].replace({'Adelie':0,'Chinstrap':1})
```

```python
data_bin_2feature_filtre['species']
```

```python
data_bin_2feature_filtre['species']=data_bin_2feature_filtre['species'].astype(int)
data_bin_2feature_filtre['species']
```

```python
Y=data_bin_2feature_filtre['species']
print("Y shape:",Y.shape)
print("Y type:",type(Y))
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42,stratify=Y)
```

```python
scaler=StandardScaler()
X_train_norm=scaler.fit_transform(X_train)
X_test_norm=scaler.transform(X_test)
X_train_norm
```

```python
plt.scatter(X_train_norm,Y_train)
plt.xlabel('bill_lenghth_mm (after standardization)')
plt.ylabel('Species (0=Adelie,1=Chinstrap)')
plt.show
```

```python
model=LogisticRegression()
model.fit(X_train_norm,Y_train)
print("a=",model.coef_)
print("b=",model.intercept_)
```

```python
X_plot = np.linspace(X_train_norm[:, 0].min(), X_train_norm[:, 0].max(), 100 ).reshape(-1, 1)
Y_plot=model.predict_proba(X_plot)[:,1]
plt.scatter(X_train_norm, Y_train, color='red')
plt.plot (X_plot, Y_plot)
plt.xlabel('bill_length_m (after standardization) - train set')
plt.ylabel("Species (0- Adelie, 1 - Chinstrap)") 
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score

p = model.predict_proba(X_test_norm)[:, 1]  # probabilité d'appartenir à la classe 1
s = 0.5  # seuil de décision
y_test_pred = (p > s).astype(int)

acc = accuracy_score(Y_test, y_test_pred)  # calcul de l'accuracy
print("Accuracy =", acc)

# Matrice de confusion
cm = confusion_matrix(Y_test, y_test_pred)

# Affichage
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.show()
```

```python
VN=cm[0,0]
FP=cm[0,1]
FN=cm[1,0]
VP=cm[1,1]

accuracy = (VP + VN) / (VP + VN + FP + FN)
precision =VP / (VP + FP)
recall =VP / (VP + FN)
specificity = VN / (VN + FP)
print ("Accuracy =",accuracy)
print("Precision =",precision)
print("Recall print =",recall)
print("Specificity =", specificity)
```
