```python
plt.figure()

plt.scatter(data['Ensoleillement_h'], data['Production_solaire_MWh'], color='orange', edgecolor='black')
plt.xlabel("Ensoleillement (h)")
plt.ylabel("Production solaire (kWh)")
plt.title("Distribution des échantillon")
plt.grid(True)
plt.show()
```

```python
X=data['Ensoleillement_h']
Y=data['Production_solaire_MWh']

w = np.random.randn(2)   # w[0] = a, w[1] = b
print(w)

n=0.01 #le nombe de pas 
N=np.size(X,0)


# ---------------------------------------------------------
# 5) Descente de gradient stochastique (SGD)
# ---------------------------------------------------------
err = np.zeros(5000)

for k in range(5000):

    # choisir aléatoirement un indice i
    i = np.random.randint(N)

    Xi = X[i]
    Yi = Y[i]

    # prédiction pour l'échantillon choisi
    Ypred_i = w[0] * Xi + w[1]

    # gradient = -2 * [Xi, 1] * (Yi - Ypred_i)
    #grad = -2 * np.array([Xi, 1]) * (Yi - Ypred_i)

    grad0 = -2 *Xi* (Yi - Ypred_i)
    grad1 = -2 *1* (Yi - Ypred_i)
    
    # mise à jour
    #w = w - n * grad
    w[0]=w[0] - n*grad0
    w[1]=w[1] - n*grad1
    
    # prédiction globale
    Ypred = w[0] * X + w[1]
    err[k]=mean_squared_error(Y,Ypred)

print("Paramètre a =", w[0])
print("Paramètre b =", w[1])

plt.figure(figsize=(4,3))
plt.scatter(X, Y)
plt.plot(X,Ypred, color='blue', label="Régression linéaire")
plt.xlabel("Ensoleillement (h)")
plt.ylabel("Production solaire (kWh)")
plt.title("Nuage de points + droite de régression")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(4,3))
plt.plot(err)
plt.title("Évolution de l’erreur (SGD)")
plt.grid(True)
plt.show()
```
