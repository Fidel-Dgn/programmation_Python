```python
# Bibliothèque mathématique Python
import math
max_pixel=255
min_pixel=1



valeur = int(input("Veuillez entrer une valeur de pixel de votre choix :"))
b=True
while(b):
    nb_niveau=int(input("Veuillez entrer le nombre de niveau :"))
    #x.type()=type(2)
    if(nb_niveau>0):
        if(math.log2(nb_niveau).is_integer()):
            b=False
        else:
            print("Le nombre de niveau doit etre un entier puissance de 2.")
    else:
        print("Le nombre de niveau doit etre un entier positif et different de zero.")

if (valeur<= max_pixel and valeur>=min_pixel):

    print("la valeur est valide")
    
    for i in range(nb_niveau):
        if(valeur<(i+1)*(256//nb_niveau)):
            valeur=i
            print("valeur sur ",nb_niveau," niveau de gris=",valeur)
            break

elif (valeur<0):
    print("la valeur fournie doit etre positif")


elif (valeur>max_pixel):

    print("la valeur fournie est trop grande")


else:
    print("Valeur non valide")
```

#### Exercice 3.2 : en partant de la copie du programme précédent modifier le code pour obtenir le traitement suivant :
On veut appliquer les mêmes traitements (binarisation, 4 niveaux, 8 niveaux de gris, N niveaux de gris) à l'image ainsi obtenue. 

A partir des codes précédemment écrits, traiter et visualiser les images issues de la transformation sélectionnée par l'utilisateur après le choix du niveau souhaité. L'image modifiée est à la base une copy de l'image initiale, et la transformation se fait sur cette copie.  


```python
import matplotlib as mp
from matplotlib import pyplot

print(tab)
long_tab = len(tab)
mon_image = []

# construction de l'image à partir de copies de la ligne tab
for i in range(long_tab):
    mon_image.append(tab.copy())
    # vérification des identifiants --> ils doivent être différents 
    print('-----', id(mon_image[i]))
    

copie_mon_image = [ligne[:] for ligne in mon_image]  # pas mon_image.copy() car cela ne change pas les identifiants des lignes de l'image du coup on ne peut pas modifier l'image sans modifier l'originale
nb_niveau=int(input("Veuillez entrer le nombre de niveau de quantification de l'image pour le traitement :"))

for i in range(long_tab):
    for j in range(long_tab):
        for k in range(nb_niveau):
            if(copie_mon_image[i][j]<(k+1)*(256//nb_niveau)):
                copie_mon_image[i][j]=k
                break
       
print('liste de liste ', type(mon_image), '\n',type(mon_image[0]), '\n', mon_image)
print('liste de liste ', type(copie_mon_image), '\n',type(copie_mon_image[0]), '\n', copie_mon_image)

# visualisation de l'image en iveau de gris
mp.pyplot.subplot(1, 2, 1)
mp.pyplot.imshow(mon_image, cmap='gray')
mp.pyplot.title('Image originale')

mp.pyplot.subplot(1, 2, 2)
mp.pyplot.imshow(copie_mon_image, cmap='gray')
mp.pyplot.title('Image quantifiée sur '+str(nb_niveau)+' niveaux de gris')


```
