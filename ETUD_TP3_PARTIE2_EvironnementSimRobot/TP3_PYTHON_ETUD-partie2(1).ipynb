#### Contexte : un robot évolue dans une pièce, comportant d'éventuels obstacles. 

    - Une pièce est représentée par ses dimensions (longueur, largeur), la position du coin haut droit (x,y), la 
        position des ouvertures (x,y,largeur_porte). 

    - Un obstacle est représenté par la position de son centre, son type (cercle ou carré) et la dimension 
        correspondante (rayon ou côté)

A l'aide de la bibliothèque turtle, on veut représenter la pièce, ses ouvertures et ses éventuels obstacles avant de faire évoluer le robot (flèche turtle) dans cet espace en évitant les obstacles présents. 

Exécuter successivement les blocs de code suivants (dans l'ordre). Analyser le code et son résultat. 

```python
#1# 
# bibliothèque nécessaires au TP
import turtle as tl
import random as rd
from enum import Enum;
```

```python
#2# 
# Prise en compte d'une pièce

# néttoyage de la table traçante
tl.clearscreen()

# Stylet en position haute - pas de traçage
tl.up()

# position du coin Haut Droit
coin_HD = (200, 200)
x_HD, y_HD = coin_HD
print('TRACE coordonnées coinHD =', x_HD, y_HD)

# déplacement au coin Haut Droit 
tl.goto(coin_HD)

# direction à prendre pou le traçage du premier côté
tl.right(90)

# Dimensions de la pièce carrée
longueur_cote = 400
angle = 90

# Données pour le traçage
tl.width(5)                  # largeur du trait
tl.color('blue')             # couleur du trait
tl.down()                    # stylet en position basse pour le traçage

# tracer les contours de la pièce
for i in range(4):
    tl.forward(longueur_cote)
    tl.right(angle)
    
tl.up()                      # retour en position haute
```

```python
#3#
# Prise en compte d'une ouverture

# caractéristiques de la porte 1
largeur_porte1 = 60
distance_porte1_mur_droit = 100
pos_porte1 = (x_HD, y_HD - distance_porte1_mur_droit)
couleur_porte1 = 'orange'
largeur_trait_porte1 = 10
print('position porte1 = ', pos_porte1)

# Traçage de la porte 1
tl.width(10)                    # largeur du trait des ouvertures
tl.up()                         # stylet en position haute  
tl.forward(distance_porte1_mur_droit)  # déplacement jusqu'à la position de la porte

tl.down()                       # stylet en position basse
tl.width(largeur_trait_porte1)  
tl.color(couleur_porte1)
tl.forward(largeur_porte1)      # traçage de la porte
tl.up()                         # retour en position haute
```

```python
#4#
# Prise en compte d'un obstable

# caractéristiques de l'obstacle
type_obstacle1 = 'cercle'
rayon_obstacle1 = 20
delta_pos_x = 150
delta_pos_y = 200
pos_obstacle1 = (x_HD - delta_pos_x , y_HD - delta_pos_y)

couleur_obstacle1 = 'green'
largeur_trait_obstacle1 = 10

print('position obstacle 1 = ', pos_obstacle1)

# Traçage de l'obstacle
tl.width(largeur_trait_obstacle1)                    # largeur du trait des ouvertures
tl.up()                                              # stylet en position haute  
tl.goto(pos_obstacle1)                               # déplacement jusqu'au la position de l'obstacle 1


tl.width(largeur_trait_obstacle1)      
tl.color(couleur_obstacle1)
if type_obstacle1 == 'cercle' : 
    tl.forward(rayon_obstacle1)
    tl.down()                           # stylet en position basse
    tl.begin_fill()
    tl.circle(rayon_obstacle1)
    tl.end_fill ()
          
tl.up()                         # retour en position haute
```

```python
#5#
# positionnement du robot à l'extérieur de la pièce
delta_robot_x = -50
delta_robot_y = distance_porte1_mur_droit + 30
pos_robot = (x_HD - delta_robot_x, y_HD - delta_robot_y)
print('TRACE position robot = ', pos_robot)
tl.goto(pos_robot)
```

```python
#6# 
#déplacement du robot dans la pièce

# caractéristique du robot
couleur_robot = 'red'
largeur_trait_robot = 3
distance_parcours1 = 100

# orientation du robot 
tl.right(angle)
tl.width(largeur_trait_robot)
tl.color(couleur_robot)
tl.down()
tl.forward(distance_parcours1)
tl.up()

#
```

```python
def positionner_robot(robot):
    if (not isinstance(robot,dict)):
        print("Erreur: le paramètre n'est pas un dictionnaire représentant un robot...")
        return
    else:
        tl.up()
        tl.goto(robot["position"])

        # Orientation absolue
        tl.setheading(robot["orientation"] * 90)

        tl.down()
        
        

```

```python
def executer_trajectoire(robot, piece):
    if not isinstance(robot, dict) or not isinstance(piece, dict):
        print("Erreur : robot ou pièce invalide.")
        return

    c = False
    marge = 5
    distance_max = 10

    for ouverture in piece["liste_ouvertures"]:
        x_robot, y_robot = robot["position"]
        x_ouv, y_ouv = ouverture["position"]
        largeur = ouverture["largeur"]
        coin = ouverture["coins_mur_ouverture"]

        if coin in (piece["coins_piece"][0], piece["coins_piece"][2]):
            y_min = min(y_ouv, y_ouv - largeur)
            y_max = max(y_ouv, y_ouv - largeur)

            if not (
                y_min + marge <= y_robot <= y_max - marge
                and abs(x_robot - x_ouv) <= distance_max
            ):
                c = True
                break
        else:
            x_min = min(x_ouv, x_ouv - largeur)
            x_max = max(x_ouv, x_ouv - largeur)

            if not (
                x_min + marge <= x_robot <= x_max - marge
                and abs(y_robot - y_ouv) <= distance_max
            ):
                c = True
                break

    if not c:
        print("Erreur : le robot n'est pas devant une ouverture.")
        return

    for trajectoire in robot["liste_parcours"]:
        for segment in trajectoire:

            # Lire la position réelle de la tortue
            positionner_robot(robot)
            position_actuelle = tuple(tl.pos())

            if not valider_position(position_actuelle, segment, 5):
                return

            depart = segment["depart"]
            direction = segment["direction"]
            distance = segment["distance"]

            # Positionner le robot au départ validé
            tl.up()
            tl.goto(depart)
            robot["position"] = depart

            # Définir l'orientation
            modifier_orientation(robot, direction)
            tl.down()
            # Exécuter le déplacement
            tl.forward(distance)

            # Enregistrer la position réelle d'arrivée
            position_arrivee = tuple(tl.pos())
            robot["position"] = position_arrivee
            segment["arrivee"] = position_arrivee

            print("Arrivée du segment :", position_arrivee)
```
